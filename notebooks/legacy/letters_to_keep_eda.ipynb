# EDA of Scrabble Valid Words Dictionary for Finding the Optimal Letters to Keep in a Scrabble Hand

```python
import pandas as pd
from collections import Counter, defaultdict
```

```python
df = pd.read_csv("word_dictionary_pd_df.csv")
```

```python
df.head(65)
```

```python
df = df.drop(['Unnamed: 0.1', 'Unnamed: 0'], axis=1)
```

```python
null_count = df.isnull().sum()

null_count
```

```python
df = df.dropna()
```

```python
seven_letters_or_more_df = df[df['legal_scrabble_words'].str.len() >= 7]
```

```python
seven_letters_or_more_df.head(10)
```

```python
# all letter counts
combined_text_all = ''.join(df['legal_scrabble_words'])

letter_counts_all = Counter(combined_text_all)

letter_counts_all = {a: b for a, b in letter_counts_all.items()}

sorted_letters_all = dict(sorted(letter_counts_all.items(), key=lambda x: x[1], reverse=True))

sorted_letters_all
```

```python
# letter counts for words with 7 or more letters
combined_text_7ormore = ''.join(seven_letters_or_more_df['legal_scrabble_words'])

letter_counts_7ormore = Counter(combined_text_7ormore)

letter_counts_7ormore = {a: b for a, b in letter_counts_7ormore.items()}

sorted_letters_7ormore = dict(sorted(letter_counts_7ormore.items(), key=lambda x: x[1], reverse=True))

sorted_letters_7ormore
```

```python
# words with 7 or more letters that contain two or more of the same letter (not consecutively)
def count_letter_frequency(df, column_name):
    letter_counts = {i: defaultdict(int) for i in range(2, 8)}
    
    for value in df[column_name]:
        char_counts = {}
        for char in value:
            if char.isalpha():
                char_counts[char] = char_counts.get(char, 0) + 1
        
        for char, count in char_counts.items():
            for i in range(2, 8):
                if count == i:
                    letter_pair = char + char
                    letter_counts[i][letter_pair] += 1
    
    all_letters = set()
    for i in range(2, 8):
        all_letters.update(letter_counts[i].keys())
    
    data = []
    for letter_pair in sorted(all_letters):
        row = {'repeated_letter': letter_pair}
        for i in range(2, 8):
            row[f'count_{i}_times'] = letter_counts[i][letter_pair]
        
        row['total_count'] = sum(letter_counts[i][letter_pair] for i in range(2, 8))
        
        data.append(row)
    
    result_df = pd.DataFrame(data).sort_values('total_count', ascending=False)
    
    return result_df

letter_distribution_7ormore = count_letter_frequency(seven_letters_or_more_df, 'legal_scrabble_words')
letter_distribution_7ormore
```
